## 6. Feature Engineering

Feature engineering creates additional variables that make the dataset easier to analyze and helps reveal business patterns that are not immediately available in the raw columns.


```python
sales_clean["year"] = sales_clean["order_date"].dt.year
sales_clean["month"] = sales_clean["order_date"].dt.month
sales_clean["month_name"] = sales_clean["order_date"].dt.month_name()
sales_clean["day_name"] = sales_clean["order_date"].dt.day_name()
```

```python
sales_clean[
    ["order_date", "year", "month", "month_name", "day_name"]
].head()
```

```python
sales_clean["discount_range"] = pd.cut(
    sales_clean["discount_amount"],
    bins=[-1, 100, 200, 300, 500, float("inf")],
    labels=["₹0–100", "₹101–200", "₹201–300", "₹301–500", "₹500+"]
)

discount_range_analysis = (
    sales_clean.groupby("discount_range", observed=False)
    .agg(
        total_sales=("net_sales", "sum"),
        total_profit=("profit", "sum"),
        orders=("order_id", "nunique")
    )
    .reset_index()
)

discount_range_analysis
```

```python
sales_clean["age_group"] = pd.cut(
    sales_clean["customer_age"],
    bins=[0, 18, 25, 35, 45, 55, 65, 100],
    labels=[
        "Under 18",
        "18–25",
        "26–35",
        "36–45",
        "46–55",
        "56–65",
        "65+"
    ]
)

age_analysis = (
    sales_clean.groupby("age_group", observed=False)
    .agg(
        customers=("customer_id", "nunique"),
        orders=("order_id", "nunique"),
        total_sales=("net_sales", "sum"),
        average_order_value=("net_sales", "mean")
    )
    .reset_index()
)

age_analysis
```

## 13. Delivery & Return Analysis

```python
delivery_analysis = (
    sales_clean.groupby("delivery_status")
    .agg(
        orders=("order_id", "nunique"),
        total_sales=("net_sales", "sum"),
        average_delivery_days=("delivery_days", "mean"),
        average_estimated_days=("estimated_delivery_days", "mean")
    )
    .reset_index()
    .sort_values("orders", ascending=False)
)

delivery_analysis
```

```python
return_analysis = (
    sales_clean.groupby("return_status")
    .agg(
        orders=("order_id", "nunique"),
        total_sales=("net_sales", "sum"),
        total_profit=("profit", "sum")
    )
    .reset_index()
    .sort_values("orders", ascending=False)
)

return_analysis
```

```python
total_records = len(sales_clean)
returned_records = sales_clean["return_status"].eq("Returned").sum()

return_rate = (returned_records / total_records) * 100

print("Total Records:", total_records)
print("Returned Records:", returned_records)
print("Return Rate:", round(return_rate, 2), "%")
```

```python
return_reasons = (
    sales_clean[sales_clean["return_status"] == "Returned"]
    ["return_reason"]
    .value_counts()
    .reset_index()
)

return_reasons.columns = ["return_reason", "returned_orders"]

return_reasons
```

**Return-rate definition:** In this project, return rate is calculated as the percentage of dataset records marked as `Returned`. It should therefore be interpreted as a record/order-level rate rather than a customer-level return rate.

## 22. Key Business Insights

Based on the completed analysis, the major business findings are:

1. **Overall scale:** The analyzed dataset contains 138,116 orders/records, 24,911 customers, approximately ₹177.13 million in net sales, and approximately ₹76.15 million in profit.
2. **Customer value:** Loyal customers contribute the largest share of sales among the customer types analyzed, showing the importance of retention.
3. **Repeat behavior:** Repeat customers contribute substantially more sales than non-repeat customers, indicating that customer retention is commercially important.
4. **Sales channels:** The Mobile App is the strongest sales channel by net sales in this dataset, followed by the Website, Marketplace, and Social Media.
5. **Regional performance:** The South region contributes the highest net sales among the regions analyzed.
6. **Marketing:** Organic Search is the highest-performing marketing channel by net sales in the analysis.
7. **Operations:** Delivery status and shipping-method analysis show that operational performance should be monitored alongside sales because delivery experience can affect customer satisfaction.
8. **Returns:** Returned orders represent a measurable portion of the dataset, making return reasons an important area for operational and customer-experience improvement.
9. **Profitability:** Customer segments should be evaluated using both sales and profit margin rather than revenue alone.
10. **Customer experience:** Ratings and review sentiment provide additional context for understanding customer experience beyond transaction value.


## 23. Business Recommendations

- Strengthen retention programs for loyal and repeat customers.
- Continue optimizing the highest-performing sales and marketing channels while monitoring channel profitability.
- Investigate major return reasons and identify operational causes that can be reduced.
- Monitor delivery performance by shipping method and warehouse.

- Evaluate discount levels using both sales and profit to avoid sacrificing profitability for volume.
- Track customer ratings and sentiment as customer-experience KPIs.
- Use customer lifetime value to prioritize high-value customer retention.


## 24. Limitations
